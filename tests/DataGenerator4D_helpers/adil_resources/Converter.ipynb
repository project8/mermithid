```python
import numpy as np
```

## Detection Efficiency

```python
de = np.load("detection_efficiency.npy", allow_pickle=True).item()
```

```python
for key, value in de.items():
    print(f"{key}: {value.shape}")
```

```python
de["theta_center_edges"]
```

```python
new_theta_center_edges = de["theta_center_edges"]
new_theta_center_edges = np.insert(new_theta_center_edges, 0, 0)
new_theta_center_edges = np.concatenate((new_theta_center_edges, 180.0 - new_theta_center_edges[-2::-1]))
new_theta_center_edges = new_theta_center_edges * np.pi / 180.0
print(new_theta_center_edges.shape)
print(new_theta_center_edges)
```

```python
new_efficiency = de["efficiency"]
print(new_efficiency[0, :, 0, 0])
print(new_efficiency.shape)

# append 0 in front of the 1st (not zeroth) axis
new_efficiency = np.insert(new_efficiency, 0, 0, axis=1)
print(new_efficiency[0, :, 0, 0])
print(new_efficiency.shape)

# mirror the 1st axis (theta_center) and append it to the original array
new_efficiency = np.concatenate((new_efficiency, new_efficiency[:, ::-1, :, :]), axis=1)
print(new_efficiency[0, :, 0, 0])
print(new_efficiency.shape)
```

```python
np.save(
    "detection_efficiency_formatted.npy",
    {
        "efficiency": new_efficiency,
        "ke_edges": de["ke_edges"],
        "theta_center_edges": new_theta_center_edges,
        "r_edges": de["r_edges"],
        "phi_edges": de["phi_edges"],
    },
)
```

## Energy Error

```python
ee = np.load("energy_error_map.npy", allow_pickle=True).item()

for key, value in ee.items():
    print(f"{key}: {value.shape}")
```

```python
ee["theta_center_edges"]
```

```python
new_theta_center_edges = ee["theta_center_edges"]
new_theta_center_edges = np.insert(new_theta_center_edges, 0, 0)
new_theta_center_edges = np.concatenate((new_theta_center_edges, 180.0 - new_theta_center_edges[-2::-1]))
new_theta_center_edges = new_theta_center_edges * np.pi / 180.0
print(new_theta_center_edges.shape)
print(new_theta_center_edges)
```

```python
new_energy_error_map = ee["energy_error_map"]
print(new_energy_error_map[0, 0, :, 0, 0])
print(new_energy_error_map.shape)

# append 0 in front of the 2nd axis
new_energy_error_map = np.insert(new_energy_error_map, 0, 0, axis=2)
print(new_energy_error_map[0, 0, :, 0, 0])
print(new_energy_error_map.shape)

# mirror the 2nd axis (theta_center) and append it to the original array
new_energy_error_map = np.concatenate((new_energy_error_map, new_energy_error_map[:, :, ::-1, :, :]), axis=2)
print(new_energy_error_map[0, 0, :, 0, 0])
print(new_energy_error_map.shape)
```

```python
np.save(
    "energy_error_map_formatted.npy",
    {
        "energy_error_map": new_energy_error_map,
        "ke_error_edges": ee["ke_error_edges"],
        "ke_start_edges": ee["ke_start_edges"],
        "theta_center_edges": new_theta_center_edges,
        "r_edges": ee["r_edges"],
        "phi_edges": ee["phi_edges"],
    },
)
```
